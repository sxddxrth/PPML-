```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import plotly.express as px

# seaborn style
sns.set()

# Correct data (same length)
Xpoints = np.array([0, 20, 120, 200, 250, 300, 350])
Ypoints = np.array([10, 80, 100, 120, 150, 180, 250])

# Matplotlib + Seaborn Line Plot
plt.figure(figsize=(8,5))
plt.plot(Xpoints, Ypoints, marker='o')
plt.title("Sample Line Plot")
plt.xlabel("X Points")
plt.ylabel("Y Points")
plt.grid(True)
plt.show()

# Plotly Interactive Plot
df = pd.DataFrame({"X": Xpoints, "Y": Ypoints})
fig = px.line(df, x="X", y="Y", markers=True, title="Interactive Plot")
fig.show()
```

```python
sns.scatterplot(x=Xpoints, y=Ypoints)
plt.title("Scatter Plot of Xpoints vs Ypoints")
plt.xlabel("X Points")
plt.ylabel("Y Points")
plt.grid(True)
plt.show()
```

```python
sns.barplot(x=np.arange(len(Ypoints)), y=Ypoints)
plt.title("Bar Plot of Ypoints")
plt.xlabel("Index")
plt.ylabel("Y Points")
plt.grid(True)
plt.show()
```
